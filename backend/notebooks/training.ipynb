```python
# Cell 1: Imports
import pandas as pd
import numpy as np
import joblib
import os
from datasets import load_dataset
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, accuracy_score
```

```python
# Cell 2: Data Loading
# Using the official AG News dataset from Hugging Face
print("Loading dataset...")
dataset = load_dataset("ag_news")
train_data = dataset['train']
test_data = dataset['test']

# Convert to DataFrame for easier manipulation
df_train = pd.DataFrame(train_data)
df_test = pd.DataFrame(test_data)
print(f"Training set: {len(df_train)} rows | Test set: {len(df_test)} rows")
```

```python
# Cell 3: Machine Learning Pipeline (Classical Approach)
# The requirement suggests TF-IDF + Simple Classifier (Naive Bayes/SVM).
# This pipeline ensures raw text input undergoes the same transformations during training and inference.
model_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(stop_words='english', max_features=10000)),
    ('clf', MultinomialNB())
])
```

```python
# Cell 4: Training
print("Starting training process...")
model_pipeline.fit(df_train['text'], df_train['label'])
print("Training completed.")
```

```python
# Cell 5: Evaluation
print("Evaluating model performance...")
predictions = model_pipeline.predict(df_test['text'])

# AG News Class Mapping: 0: World, 1: Sports, 2: Business, 3: Sci/Tech
class_names = ['World', 'Sports', 'Business', 'Sci/Tech']

acc = accuracy_score(df_test['label'], predictions)
print(f"Final Accuracy: {acc:.4f}")
print("\nDetailed Classification Report:")
print(classification_report(df_test['label'], predictions, target_names=class_names))
```

```python
# Cell 6: Model Persistence
# Relative path: exit 'notebooks', enter 'model'
output_path = '../model/trained_model.pkl'

# Ensure directory exists
os.makedirs(os.path.dirname(output_path), exist_ok=True)

joblib.dump(model_pipeline, output_path)
print(f"Model successfully saved to: {output_path}")
```
